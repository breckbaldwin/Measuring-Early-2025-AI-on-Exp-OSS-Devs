```python
import pandas as pd
data_df = pd.read_csv('data_complete.csv')
data_df['ai_treatment'] = data_df['ai_treatment'].astype(int)

for cond in [0, 1]:
    cond_df = data_df[data_df['ai_treatment']==cond]
    time = 'initial_implementation_time'
    print(f'AI: {cond} {time}', 
          'mean:', cond_df[time].mean(),
          'std:', cond_df[time].std(),
          'min:', cond_df[time].min(),
          'max:', cond_df[time].max(),
          )



for dev in data_df['dev_id'].unique():
    dev_df = data_df[data_df['dev_id']==dev]
    print(f'Dev: {dev}')

    for cond in [0, 1]:
        cond_df = dev_df[dev_df['ai_treatment']==cond]
        print(f'AI: {cond} {time}', 
              'mean:', cond_df[time].mean(),
              'std:', cond_df[time].std(),
              'min:', cond_df[time].min(),
              'max:', cond_df[time].max(),)

    
```

```python
# Create diff_df that subtracts means of ai_treatment 0 from ai_treatment 1
# Calculate means for each treatment group
treatment_0_mean = data_df[data_df["ai_treatment"] == 0]["initial_implementation_time"].mean()
treatment_1_mean = data_df[data_df["ai_treatment"] == 1]["initial_implementation_time"].mean()

# Calculate the difference (treatment 1 - treatment 0)
mean_difference = treatment_1_mean - treatment_0_mean

# Create diff_df with the results
diff_df = pd.DataFrame({
    "metric": ["initial_implementation_time"],
    "treatment_0_mean": [treatment_0_mean],
    "treatment_1_mean": [treatment_1_mean],
    "difference": [mean_difference],
    "difference_description": [f"AI treatment 1 is {mean_difference:.2f} units{\"higher\" if mean_difference > 0 else \"lower\"} than treatment 0"]
})

display(diff_df)

# Also show the raw means for comparison
print(f"\nRaw means:")
print(f"AI Treatment 0: {treatment_0_mean:.2f}")
print(f"AI Treatment 1: {treatment_1_mean:.2f}")
print(f"Difference (1 - 0): {mean_difference:.2f}")
```
